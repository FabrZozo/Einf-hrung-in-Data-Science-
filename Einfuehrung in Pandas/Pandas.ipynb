# Gruppierung

```python
df = pd.DataFrame({
    "Name":   ["Chris", "Clara", "Fabrice", "Lina", "Cedric", "Anna"],
    "Klasse": ["A", "A", "B", "B", "A", "B"],
    "Maths":  [17.0, 10.0, 13.0, 10.5, 13.5, 15.0],
    "Physik": [15.0, 8.0, 10.5, 17.0, 13.5, 12.0]
})
df
```

```python
df.groupby("Klasse").max(numeric_only=True)
```

```python
df.groupby("Klasse")["Maths"].mean()
```

```python
df.groupby("Klasse")["Maths"].agg(["mean", "max", "min", "count"])
```

```python
df.groupby("Klasse").count()
```
