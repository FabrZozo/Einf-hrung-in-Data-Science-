```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

```python
data= make_blobs(n_samples=200, n_features=2, centers=3, cluster_std=1.5,random_state=100)
```

```python
data
```

```python
data1=data[0]
data1
```

```python
plt.scatter(data1[:,0],data1[:,1])
```

```python
plt.scatter(data1[:,0],data1[:,1],c=data[1],cmap="viridis")
```

```python
dataMeans= KMeans(n_clusters=3)
```

```python
dataMeans.fit(data1)
```

```python
dataMeans.cluster_centers_
```

```python
dataMeans.labels_
```

```python
plt.scatter(data1[:,0],data1[:,1],c=dataMeans.labels_,cmap="viridis")
```
